```python
import pandas as pd
import numpy as np

athletes = pd.read_csv("athlete_events.csv")
nocs = pd.read_csv("noc_regions.csv")
```

```python
athletes.head()
athletes.info()
```

```python
df = athletes.merge(nocs, on="NOC", how="left")
```

```python
df.head()
```

```python
df["Medal_Won"] = df["Medal"].notna().astype(int)
```

```python
df["Gold"] = (df["Medal"] == "Gold").astype(int)
df["Silver"] = (df["Medal"] == "Silver").astype(int)
df["Bronze"] = (df["Medal"] == "Bronze").astype(int)
```

```python
df.head()
```

```python
df[["Age", "Height", "Weight"]].isna().mean()
```

```python
df = df[df["Season"] == "Summer"]
```

```python
df.head()
```

```python
athlete_counts = (
    df.groupby(["region", "Sport"])["ID"]
    .nunique()
    .reset_index(name="Athletes_Sent")
)
athlete_counts.head()
```

```python
medal_counts = (
    df[df["Medal_Won"] == 1]
    .groupby(["region", "Sport"])["Medal_Won"]
    .sum()
    .reset_index(name="Medals_Won")
)
medal_counts.head()
```

```python
efficiency = athlete_counts.merge(
    medal_counts,
    on=["region", "Sport"],
    how="left"
).fillna(0)

efficiency["Medal_Efficiency"] = (
    efficiency["Medals_Won"] / efficiency["Athletes_Sent"]
)

efficiency.head()
```

```python
df = df.rename(columns={"region": "Country"})
athlete_counts = athlete_counts.rename(columns={"region": "Country"})
medal_counts = medal_counts.rename(columns={"region": "Country"})
efficiency = efficiency.rename(columns={"region": "Country"})
```

```python
sport_efficiency = (
    efficiency.groupby("Sport")
    .agg(
        Total_Athletes=("Athletes_Sent", "sum"),
        Total_Medals=("Medals_Won", "sum")
    )
)

sport_efficiency["Medal_Efficiency"] = (
    sport_efficiency["Total_Medals"] / sport_efficiency["Total_Athletes"]
)

sport_efficiency.sort_values("Medal_Efficiency", ascending=False).head(10)
```

```python
age_by_sport = (
    df.groupby(["Sport", "Medal_Won"])["Age"]
    .median()
    .unstack()
)
age_by_sport.head()
```

```python
sport_participation = (
    df.groupby("Sport")["ID"]
    .nunique()
    .sort_values(ascending=False)
)

sport_participation.head(10)
```

```python
countries_per_sport = (
    df.groupby("Sport")["Country"]
    .nunique()
    .sort_values(ascending=False)
)

countries_per_sport.head(10)
```

```python
sport_efficiency.sort_values("Medal_Efficiency", ascending=False).head(15)
```

```python
medals_by_country_sport = (
    df[df["Medal_Won"] == 1]
    .groupby(["Country", "Sport"])["Medal_Won"]
    .sum()
    .reset_index()
)

total_medals_by_country = (
    medals_by_country_sport
    .groupby("Country")["Medal_Won"]
    .sum()
    .reset_index(name="Total_Medals")
)

country_sport_share = medals_by_country_sport.merge(
    total_medals_by_country,
    on="Country"
)

country_sport_share["Share_of_Medals"] = (
    country_sport_share["Medal_Won"] /
    country_sport_share["Total_Medals"]
)
country_sport_share.head()
```

```python
medal_ages = (
    df[(df["Medal_Won"] == 1) & (df["Age"].notna())]
    [["Sport", "Age"]]
)
```

```python
age_iqr_by_sport = (
    medal_ages
    .groupby("Sport")["Age"]
    .agg(
        Median_Age="median",
        Q1_Age=lambda x: x.quantile(0.25),
        Q3_Age=lambda x: x.quantile(0.75),
        Medal_Count="count"
    )
    .reset_index()
)

age_iqr_by_sport["IQR"] = (
    age_iqr_by_sport["Q3_Age"] - age_iqr_by_sport["Q1_Age"]
)
age_iqr_by_sport.head()
```

```python
age_iqr_by_sport = age_iqr_by_sport[
    age_iqr_by_sport["Medal_Count"] >= 100
]
```

```python
age_iqr_by_sport.sort_values("Median_Age").head(10)
```
